```python
# [APPENDIX TASK 1] EQUIPMENT DIAGNOSTIC SYSTEM
LAST_NAME = "DE JESUS"
SEED_NUM = 3
FAVORITE_ARTIST = "BEN&BEN"

def log_decorator(func):
    def wrapper(*args, **kwargs):
        print("Execution Log:")
        print(f"-> Running function: {func.__name__} using artist {FAVORITE_ARTIST}")
        return func(*args, **kwargs)
    return wrapper

def validate_data(readings):
    validations = []
    for r in readings:
        try:
            if r < 0:
                raise ValueError("Negative entry")
            validations.append("VALID")
        except ValueError:
            validations.append("INVALID")
    return validations

def compute_and_classify(readings, validations, seed):
    diagnostics = []
    for r, v in zip(readings, validations):
        if v == "VALID":
            processed = r * seed
            status = "CRITICAL" if processed > 50 else "NORMAL"
            diagnostics.append((r, "VALID", processed, status))
        else:
            diagnostics.append((r, "INVALID", 0, "ERROR"))
    return diagnostics

@log_decorator
def run_diagnostic_system():
    # Student specific raw equipment readings based on inputs
    raw_data = [len(LAST_NAME) * 2, SEED_NUM * 15, -5, len(FAVORITE_ARTIST) * 5]
    
    print("Generated Equipment Data:", raw_data)
    
    validations = validate_data(raw_data)
    print("Validation Results:", validations)
    
    diagnostics = compute_and_classify(raw_data, validations, SEED_NUM)
    print("Diagnostic Results:", diagnostics)
    
    final_summary = f"System check completed for {LAST_NAME}. Total items checked: {len(raw_data)}."
    print("Final Output:", final_summary)

# Run the program in Jupyter
run_diagnostic_system()
```

```python
# [APPENDIX TASK 2] RECURSIVE FAULT TRACE
LAST_NAME = "DE JESUS"
SEED_NUM = 3
FAVORITE_ARTIST = "BEN&BEN"

# 1. Generate student-specific initial fault code
initial_fault_code = (len(LAST_NAME) * 5) + (len(FAVORITE_ARTIST) * 2) + SEED_NUM

# Variables to track recursion steps and counts
fault_trace_list = []
recursive_call_counter = 0

def trace_fault(code):
    global recursive_call_counter
    recursive_call_counter += 1
    fault_trace_list.append(code)
    
    # 3. Base condition (stops when code reaches 1 or below)
    if code <= 1:
        return "STABLE"
    
    # 2. Recursive step (reducing by SEED_NUM)
    return trace_fault(code - SEED_NUM)

# 5. Execution Log and Display
print("Generated Fault Data:", initial_fault_code)
print(f"Execution Log -> Initiating recursive fault trace for {LAST_NAME} (Artist: {FAVORITE_ARTIST})")

final_status = trace_fault(initial_fault_code)

print("Recursive Trace:", fault_trace_list)
print("Number of Recursive Calls:", recursive_call_counter)
print(f"Final Output: Fault tracking completed for {LAST_NAME}. Status: {final_status}")
```

```python
# [APPENDIX TASK 3] INTELLIGENT EQUIPMENT MONITORING PIPELINE
LAST_NAME = "DE JESUS"
SEED_NUM = 3
FAVORITE_ARTIST = "BEN&BEN"

# 1. Student-Specific Inputs Display
print("Student-Specific Inputs:")
print(f"LAST_NAME: {LAST_NAME}, SEED_NUM: {SEED_NUM}, FAVORITE_ARTIST: {FAVORITE_ARTIST}")

# 5. Decorator to monitor major processing function
def monitor_pipeline(func):
    def wrapper(*args, **kwargs):
        print("Execution Log:")
        print(f"-> Initializing telemetry monitoring pipeline for {LAST_NAME}")
        result = func(*args, **kwargs)
        print("-> Pipeline execution completed successfully.")
        return result
    return wrapper

# 3. Generator to produce and process telemetry values efficiently
def telemetry_generator():
    base_values = [15, 30, -10, 45, 60, -5, 25]
    for val in base_values:
        # Incorporating student seed and name/artist length variations
        adjusted = val + SEED_NUM
        yield adjusted

# 7. Recursive function to trace/analyze detected abnormal conditions
def recursive_fault_resolver(val):
    if val <= 5:
        return "RESOLVED"
    return recursive_fault_resolver(val - SEED_NUM)

@monitor_pipeline
def run_pipeline():
    # 2. Program organized structure
    raw_stream = list(telemetry_generator())
    print("Generated Telemetry Data:", raw_stream)
    
    valid_results = []
    invalid_count = 0
    processed_readings = []
    detected_abnormals = []
    recursive_analysis_results = []
    
    # 4. Lambda function to filter/transform data (e.g., checking threshold)
    is_normal = lambda x: x <= 50
    
    for val in raw_stream:
        try:
            # 6. Exception handling for invalid or unexpected telemetry values
            if val < 0:
                raise ValueError(f"Negative telemetry detected: {val}")
            
            valid_results.append(val)
            
            if is_normal(val):
                processed_readings.append((val, "NORMAL"))
            else:
                detected_abnormals.append(val)
                # Apply recursion for abnormal conditions
                res = recursive_fault_resolver(val)
                recursive_analysis_results.append((val, res))
                processed_readings.append((val, "ABNORMAL"))
                
        except ValueError as e:
            invalid_count += 1
            valid_results.append(f"INVALID ({e})")
            
    print("Valid/Invalid Results:", valid_results)
    print("Processed Results:", processed_readings)
    print("Recursive Analysis:", recursive_analysis_results)
    
    print("Final Diagnostic Summary:")
    print(f"Total Processed Readings: {len(raw_stream)}")
    print(f"Valid Readings: {len(raw_stream) - invalid_count}, Invalid Readings: {invalid_count}")
    print(f"Detected Abnormal Conditions: {len(detected_abnormals)}")
    
    # 8 & 9. Final Output
    overall_status = "CRITICAL" if len(detected_abnormals) > 2 else "OPTIMAL"
    print(f"Final Output: System pipeline finished for {LAST_NAME}. Overall Equipment Status: {overall_status}")

# Run Task 3 Pipeline
run_pipeline()
```
